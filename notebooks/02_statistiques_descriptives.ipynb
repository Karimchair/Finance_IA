# Notebook 02 — Statistiques descriptives

Objectif : transformer les prix en rendements, puis calculer des indicateurs simples pour chaque action.

À la fin de ce notebook, on obtient :
- `data/obtenue/rendements_journaliers.csv`
- `data/obtenue/indicateurs_actions.csv`
- plusieurs graphiques dans `figures/`

```python
# TODO faire par karim
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

os.makedirs("../data/obtenue", exist_ok=True)
os.makedirs("../figures", exist_ok=True)

prix = pd.read_csv("../data/brut/prix_actions.csv", index_col=0, parse_dates=True)
prix.head()
```

## 1. Calcul des rendements journaliers

Formule :

`rendement_journalier = (prix_du_jour / prix_de_la_veille) - 1`

Exemple : si une action passe de 100€ à 102€, son rendement est 2%.

```python
# TODO faire par karim
rendements = prix.pct_change().dropna(how="all")

# Pour éviter les problèmes, on remplace les éventuelles lignes avec trop de valeurs manquantes.
rendements = rendements.dropna()

rendements.to_csv("../data/obtenue/rendements_journaliers.csv")

rendements.head()
```

## 2. Construction du tableau des indicateurs

Chaque action devient une ligne du tableau.

Indicateurs utilisés :
- rendement moyen annuel
- volatilité annuelle
- rendement minimum journalier
- rendement maximum journalier
- corrélation avec le CAC 40

```python
# TODO faire par karim
tickers = [
    "TTE.PA", "SU.PA", "MC.PA", "AI.PA", "SAN.PA",
    "AIR.PA", "SAF.PA", "BNP.PA", "OR.PA", "CS.PA",
    "DG.PA", "EL.PA", "RMS.PA", "KER.PA", "ACA.PA"
]

indice = "^FCHI"

indicateurs = pd.DataFrame(index=tickers)

# En finance, on prend souvent environ 252 jours de bourse par an.
jours_bourse = 252

indicateurs["rendement_moyen_annuel"] = rendements[tickers].mean() * jours_bourse
indicateurs["volatilite_annuelle"] = rendements[tickers].std() * np.sqrt(jours_bourse)
indicateurs["rendement_min_journalier"] = rendements[tickers].min()
indicateurs["rendement_max_journalier"] = rendements[tickers].max()

for action in tickers:
    indicateurs.loc[action, "correlation_CAC40"] = rendements[action].corr(rendements[indice]) # c'est correlation de chaque entreprise avec CAC40

indicateurs.to_csv("../data/obtenue/indicateurs_actions.csv")

indicateurs
```

## 3. Interprétation simple des indicateurs

- Un rendement moyen annuel élevé indique que l'action a eu une bonne performance moyenne sur la période.
- Une volatilité élevée indique que l'action est instable, donc plus risquée.
- Un rendement minimum très négatif indique une forte baisse observée sur une journée.
- La corrélation avec le CAC 40 montre si l'action suit fortement le marché français.

```python
# TODO faire par karim
# Affichage en pourcentage c'est facile à lire
indicateurs_pourcentage = indicateurs.copy()
for col in indicateurs_pourcentage.columns:
    indicateurs_pourcentage[col] = indicateurs_pourcentage[col] * 100

indicateurs_pourcentage.round(2)
```
