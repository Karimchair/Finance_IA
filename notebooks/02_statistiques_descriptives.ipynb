## 1. Calcul des rendements journaliers

Formule :

`rendement_journalier = (prix_du_jour / prix_de_la_veille) - 1`

Exemple : si une action passe de 100€ à 102€, son rendement est 2%.

```python
# TODO faire par karim
rendements = prix.pct_change().dropna(how="all")

# Pour éviter les problèmes, on remplace les éventuelles lignes avec trop de valeurs manquantes.
rendements = rendements.dropna()

rendements.to_csv("../data/obtenue/rendements_journaliers.csv")

rendements.head()
```
